```python
import pandas as pd

# Loading the starter CSV file
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

# Display the first few rows and basic info to understand the data
print("First 5 rows of the dataset:")
display(df.head())
print("\nDataFrame Info:")
df.info()
```

The goal is to identify pages that are visible (high impressions) but under-capture clicks or engagement (low CTR). Let's analyze pages with the highest impressions but with lowest CTR and sort them by highest visibility and lowest CTR.

```python
import numpy as np

# Keeping only pages that were actually visible in search
visible_pages = df[df["impressions_90d"] > 0].copy()

# Replace missing clicks with 0 (if any)
visible_pages["clicks_90d"] = visible_pages["clicks_90d"].fillna(0)

# Calculating CTR safely
visible_pages["ctr"] = (
    visible_pages["clicks_90d"] /
    visible_pages["impressions_90d"]
) * 100

# Remove any remaining NaN or infinite CTR values
visible_pages = visible_pages.replace([np.inf, -np.inf], np.nan)
visible_pages = visible_pages.dropna(subset=["ctr"])

# Sort by high visibility and low CTR
underperforming_pages = visible_pages.sort_values(
    by=["impressions_90d", "ctr"],
    ascending=[False, True]
)
# printing those pages with lowest ctr and highest impressions .ie that are the pages that needs most attention
print("Top 5 highly visible pages with the lowest CTR:")
display(underperforming_pages.head())
```

```python
# Let's extract 2-3 real numbers to explain why this lane is worth considering.

# Number 1: The highest impressions and clicks count in the dataset
highest_impressions = df['impressions_90d'].max()
highest_clicks = df['clicks_90d'].max()
print(f"1. The highest number of impressions for a single page is: {highest_impressions:,} and the highest number of clicks is: {highest_clicks:,} showing a significant difference.")

# Number 2: The average CTR across all pages
# We'll use a weighted average if possible, but for simplicity, let's take a simple average of existing CTRs
# For pages with 0 impressions, CTR is 0, so including them in average might skew results.
# Let's consider only pages with impressions > 0 for average CTR
average_ctr_with_impressions = df[df['impressions_90d'] > 0]['ctr'].mean()
print(f"2. The average CTR for pages with at least one impression is approximately: {average_ctr_with_impressions:.2f}%")

# Number 3: The CTR of a highly visible but underperforming page
# Let's take the first page from our 'underperforming_pages' list (highest impressions, lowest CTR)
if not underperforming_pages.empty:
    example_page = underperforming_pages.iloc[0]
    print(f"3. An example of a highly visible page has {example_page['impressions_90d']:,} impressions but only a CTR of {example_page['ctr']:.2f}%,")
    print(f"   resulting in {example_page['clicks_90d']:,} clicks. This indicates significant untapped potential.")
else:
    print("Could not find underperforming pages to display example.")
```

These numbers demonstrate that there are pages with very high visibility (impressions) that are not translating into proportional clicks (low CTR). This confirms the problem statement of my chosen lane, indicating a clear opportunity to improve performance by reviewing such pages.
